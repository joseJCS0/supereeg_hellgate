```python
import numpy as np
import matplotlib.pyplot as plt
import supereeg as se
import glob
import os
%matplotlib
```

```python
results_path = "/home/josecs/miniconda3/envs/supereeg_env/data/npz_cluster/results/raw_recon"
```

```python
within_files = glob.glob(os.path.join(results_path, "*within*"))
```

```python
all_files = glob.glob(os.path.join(results_path, "*"))
across_files = [f for f in all_files if "within" not in os.path.basename(f)]
```

```python
within_data = np.zeros(len(within_files))
for i in range(len(within_files)):
    load = np.load(within_files[i])
    within_data[i] = load["corrs"]
```

```python
across_data = np.zeros(len(across_files))
for i in range(len(across_files)):
    load = np.load(across_files[i])
    across_data[i] = load["corrs"]
```

```python
fig, ax = plt.subplots(1,figsize=(10,6))

ax.hist(within_data,bins=200,color="grey",alpha=0.7,label="Within")
ax.hist(across_data,bins=200,color="black",alpha=0.7,label="Across")

ax.set_xlabel("Correlation")
ax.set_ylabel("Concentration")
ax.legend()
ax.grid()
```
